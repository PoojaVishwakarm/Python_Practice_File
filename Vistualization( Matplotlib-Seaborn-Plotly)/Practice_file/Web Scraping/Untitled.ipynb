```python
import pandas as pd
import requests

# step 1: URL
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

# step 2: Get the HTML page with a user-agent header
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)

# step 3: Read tables from the HTML content
tables = pd.read_html(response.text)

# step 4: Check how many tables found
print(f"Number of tables found: {len(tables)}")

# step 5: Save the first table to CSV
tables[0].to_csv("world_population_11052025.csv", index=False)
print("Saved CSV successfully.")
```

```python
import pandas as pd
import requests
from io import StringIO

url = "https://www.worldometers.info/world-population/population-by-country/"
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)
response.raise_for_status()

# Force UTF-8 decoding
response.encoding = "utf-8"

# Wrap in StringIO to avoid FutureWarning
tables = pd.read_html(StringIO(response.text))

df = tables[0]

# Preview first few rows
print(df.head())

# Save clean CSV
df.to_csv("world_population_by_country.csv", index=False, encoding="utf-8")
print("CSV saved successfully with proper encoding!")
```
