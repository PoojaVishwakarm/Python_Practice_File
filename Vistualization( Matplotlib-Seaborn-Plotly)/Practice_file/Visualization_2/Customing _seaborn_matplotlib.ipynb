## Customizing Plots in Seaborn and Matplotlib

## Loading Libraries and Data 

```python
# Sample data 
# Load the sample diamond data 
dia=pd.read_csv('diamonds.csv')
print(f" Number of row s before sampling:{dia.shape[0]}")
print(f"Number of columns befoe dsmpling :{dia.shape[1]}")
print(f"{50*'-'}")
dia=dia.sample(50000)
print(f" Number of row s before sampling:{dia.shape[0]}")
print(f"Number of columns befoe dsmpling :{dia.shape[1]}")

# Display th first few rows of the datase

dia.head()
```

# 1.Setting Titles

```python
sns.scatterplot(x='carat',y='price',data=dia)
plt.title('Scatter Plot of carat vs price') # Setting the title
plt.show()
```

# 2.Setting Axes Labels

```python
sns.scatterplot(x='carat',y='price',data=dia)
plt.xlabel('Carat Weight') # Label for x-axis
plt.ylabel('Price (in USD)')
plt.show()
```

# 3.Customizing the legend


```python
sns.scatterplot(x='carat',y='price',hue='cut',data=dia)
```

```python
sns.scatterplot(x='carat',y='price',data=dia,hue='cut')
plt.legend(title='diamond cut',loc='upper right')
plt.show()
```

# 4. Customizing Ticks

```python
sns.histplot(dia['price'],bins=10,edgecolor='black')
plt.xticks(rotation=45)
plt.yticks(fontsize=10)
plt.show()
```

# 5.Adding a Grid

```python
sns.set_style('whitegrid')
sns.scatterplot(x='carat',y='price',data=dia)
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

# 6 Customizing Spines

```python
sns.boxplot(x='cut',y='price',data=dia)
sns.despine() # Remove the top and right spines for a cleaner look
plt.show()
```

# 7.Setting Figure Size

```python

plt.figure(figsize=(10,6))
sns.barplot(x='cut',y='price',data=dia )
plt.title("Bar plot of price by cut" )
plt.show()
          
```

# 8.Creating Subplots

```python
fig,axes=plt.subplots(1,2,figsize=(12,4))
sns.histplot(dia['carat'],bins=10,edgecolor='black',ax=axes[0],kde=True)
axes[0].set_title('histogram of carat')
sns.histplot(dia['price'],bins=10,edgecolor='black',ax=axes[1],kde=True)
axes[1].set_title('histogram of price')
plt.show()
```

# 9. Adding Annotations

```python
dia.reset_index(drop=True, inplace=True)
sns.scatterplot(x='carat', y='price', data=dia)
plt.title("Scatter Plot with Annotations")

for i in range(len(dia)):
    plt.annotate(f"${dia['price'][i]}", (dia['carat'][i], dia['price'][i]), fontsize=8, alpha=0.4)

plt.show()
```

# 10. Customizing color palette

```python
sns.set_palette('muted') # Muted deep pastel, bright, dark, colorblind
sns.boxplot(x='cut',y='price',data=dia)
plt.title("Box Plot with muted color Palette")
plt.show()
```

```python
sns.set_palette('pastel') # Muted deep pastel, bright, dark, colorblind
sns.boxplot(x='cut',y='price',data=dia)
plt.title("Box Plot with muted color Palette")
plt.show()
```
